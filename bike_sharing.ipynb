```python
import os
import urllib.request
import zipfile
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Load the dataset. If the CSV is not in the current folder, download it from UCI.
csv_file = 'SeoulBikeData.csv'
if not os.path.exists(csv_file):
    url = 'https://archive.ics.uci.edu/static/public/560/seoul%2Bbike%2Bsharing%2Bdemand.zip'
    zip_file = 'seoul_bike_dataset.zip'
    urllib.request.urlretrieve(url, zip_file)
    with zipfile.ZipFile(zip_file, 'r') as z:
        z.extractall('.')
    if not os.path.exists(csv_file):
        raise FileNotFoundError('SeoulBikeData.csv was not found after downloading the UCI dataset.')

df = pd.read_csv(csv_file, encoding='unicode_escape', parse_dates=[0], dayfirst=True)
df.head(5)
```

```python
df['Spring'] = (df['Seasons'] == 'Spring').astype(int)
df['Summer'] = (df['Seasons'] == 'Summer').astype(int)
df['Winter'] = (df['Seasons'] == 'Winter').astype(int)
df['Autumn'] = (df['Seasons'] == 'Autumn').astype(int)
df['Holiday'] = (df['Holiday'] == 'Holiday').astype(int)
df['Functioning Day'] = (df['Functioning Day'] == 'Yes').astype(int)

df.drop(columns=['Seasons', 'Date'], inplace=True)
```

```python
df.head(10)
```

```python
df = df.rename(columns={
    'Temperature(°C)': 'Temperature',
    'Humidity(%)': 'Humidity',
    'Rainfall(mm)': 'Rainfall',
    'Snowfall (cm)': 'Snowfall',
    'Wind speed (m/s)': 'Wind speed',
    'Visibility (10m)': 'Visibility',
    'Solar Radiation (MJ/m2)': 'Radiation',
    'Dew point temperature(°C)': 'Dew point temperature'
})
```

```python
df.describe()
```

```python
data = np.array(df)
```

```python
y = data[:, 0]
X = data[:, 1:]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
```

```python
y.shape, X.shape, X_train.shape, X_test.shape
```

```python
# Random Forest Regression
Model = RandomForestRegressor(n_estimators=100, random_state=42)
Model.fit(X_train, y_train)
rf_pred = Model.predict(X_test)

rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_r2 = r2_score(y_test, rf_pred)

print('Random Forest Regression')
print(f'MAE : {rf_mae:.2f}')
print(f'RMSE: {rf_rmse:.2f}')
print(f'R²  : {rf_r2:.4f}')
```

```python
# Linear Regression
Model2 = LinearRegression()
Model2.fit(X_train, y_train)
lr_pred = Model2.predict(X_test)

lr_mae = mean_absolute_error(y_test, lr_pred)
lr_rmse = np.sqrt(mean_squared_error(y_test, lr_pred))
lr_r2 = r2_score(y_test, lr_pred)

print('Linear Regression')
print(f'MAE : {lr_mae:.2f}')
print(f'RMSE: {lr_rmse:.2f}')
print(f'R²  : {lr_r2:.4f}')
```
